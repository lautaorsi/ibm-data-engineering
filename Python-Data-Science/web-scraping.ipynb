Las importamos al código y creamos un "archivo HTML" simulado, sería el equivalente a lo que obtenemos en el campo `text` al realizar una GET request de una página 

```python
from bs4 import BeautifulSoup 
import requests  

html="%%html<!DOCTYPE html><html><head><title>Artistas</title></head><body><h3 class= 'c1 c2 c3'>Milo J</h3><p> Oyentes Mensuales: 2,500,000 </p><h3> Wasia Project</h3><p> Oyentes Mensuales: 1,000, 000 </p><h3> Serú Girán </h3><p id='top'> Oyentes Mensuakes: 3,200, 000</p></body></html>"
```

## BeautifulSoup

Esta libreria nos permite mappear el documento HTML a un arbol recursivo navegable, pasemosle nuestro documento y veamos el esquema del objeto 

```python
soup = BeautifulSoup(html, 'html.parser')

print(soup.prettify())
```

### Tags

En primera instancia tenemos acceso a cualquier tag del documento, siendo estos equivalentes a los tags de un documento HTML (`body`, `h3`, `b`, `title`, `head`), en caso de tener varios elementos iguales obtenemos la primer ocurrencia del tag.

```python
print((soup.body).prettify())
```

```python
soup.title
```

A su vez como mencionamos anteriormente estos objetos forman parte de un arbol navegable, por lo tanto podemos acceder al respectivo padre, hijos y hermanos

```python
#Padre del body es el archivo html
soup.body.parent
```

### Atributos

Además estos objetos tienen sus atributos diccionarizados 

```python
first_h3 = soup.h3

print(first_h3.attrs)

print(first_h3["class"])
```

### FindAll

Como dijimos antes el metodo de acceso a tags nos retorna la instancia de nivel más alto en el arbol, si por ejemplo quisieramos obtener todas las instancias de una clase de objeto podemos utilizar le metodo `findAll`

```python
h3_list = soup.find_all("h3")

print(h3_list)
```

### Filtros

Lógicamente en escenarios reales las páginas web tienden a ser un poco más complejas que 3 headers, por lo tanto es útil filtrar los tags para matchearlos con patrones específicos.

```python
filtered_h3_list = soup.find_all("h3", class_="c1")

print(filtered_h3_list)

filtered_p_list = soup.find_all("p", id=(lambda value: value != "top"))

print(filtered_p_list)
```
